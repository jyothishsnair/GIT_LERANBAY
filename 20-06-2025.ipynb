Let's see how we can perform five number summary analysis in Python. There are various approaches or techniques one may use depending on their needs.

Consider the following data representing the number of passengers on specific flights between Delhi and Hyderabad over the last 33 days:
{128, 121, 134, 136, 136, 118, 123, 109, 120, 116, 125, 128, 121, 129, 130, 127, 119, 114, 134, 110, 136, 134, 125, 128, 123, 128, 133, 132, 136, 134, 129, 132, 145, 178}

Analyze the descriptive statistics to gain a basic understanding of the data.


```python
#Importing Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Ignore Warnings
import warnings
warnings.filterwarnings('ignore')
```

 Used in Python to suppress warning messages that would otherwise appear during code execution.


```python
Data=[128,121,134,136,136,118,123,109,120,116,125,128,121,129,130,127,119,114,134,110,136,134,125,128,123,128,133,132,136,134,129,132,145,178]
```

```python
print(f'Data type of the give data set is {type(Data)}.') # To understand the type of data set.
```

```python
Data = np.array(Data) #Convert the list of an array
Data
```

```python
Passengers = pd.DataFrame(Data,columns=['Number of Passengers']) # This code will help us to understand how we can create a dataframe.
```

```python
Passengers.head() #To get the Five 5 sample data
```

```python

print(f'\nFIVE NUMBER SUMMARY\n')
print(f'This data has {Passengers.shape[0]} rows and {Passengers.shape[1]} colums.')
print("Mean : ",round(Passengers['Number of Passengers'].mean(),2))
print("Min : ",round(Passengers['Number of Passengers'].min(),2))
print("25% : ",round(Passengers['Number of Passengers'].quantile(0.25),2))
print("50% : ",round(Passengers['Number of Passengers'].quantile(0.50),2))
print("75% : ",round(Passengers['Number of Passengers'].quantile(0.75),2))
print("Max : ",round(Passengers['Number of Passengers'].max(),2))
```

```python
Passengers.describe() #We can use describe method to understand the five number summary
```

```python
sns.boxplot(Passengers['Number of Passengers'], orient='h')
plt.show()
```

```python
fig =plt.figure(figsize = (6,5))
# Plot the boxplot
sns.boxplot(Passengers['Number of Passengers'], orient='h')
plt.title("Boxplot with Median and Quartiles")

# Get quartiles
q1 = Passengers['Number of Passengers'].quantile(0.25)
q2 = Passengers['Number of Passengers'].median()
q3 = Passengers['Number of Passengers'].quantile(0.75)

# Add annotations
plt.axvline(x=q1, color='orange', linestyle='--', label='Q1')
plt.axvline(x=q2, color='red', linestyle='--', label='Median (Q2)')
plt.axvline(x=q3, color='green', linestyle='--', label='Q3')
plt.legend()
plt.show()
```
